```python
class Dog:
    """model dogs- docstring"""
    def __init__(self, name:str, age:int, weight:float):
        self.name = name
        self.age = age 
        self.weight = weight
    def sit(self):
        print(f'{self.name} is sitting')
    def rollover(self):
        print(f"{self.name} is rolling")
```

```python
my_dog = Dog('Xoai',1, 3.1)
```

```python
my_dog.name = "Xoài"
my_dog.name
```

```python
my_dog.rollover()
my_dog.sit()
```

```python
class Car:
    """Car docstring"""
    def __init__(self, brand, model,color):
        self.brand = brand
        self.model = model
        self.color = color
    def repaint(self):
        print(f"the car is paiting, color: {self.color}")
```
