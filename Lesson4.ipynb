```python
def is_prime(n:int)->bool:
    '''check prime'''
    if n<=1:
        return False
    else:
        for d in range(2,int(n**0.5)+1):
            if n%d==0:
                return False
    return True 
```

```python
is_prime(213073013129387)
```

```python
is_prime(9)
```
