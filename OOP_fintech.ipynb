## Giới thiệu

Chào mừng các bạn đến với bài giảng về Lập trình Hướng đối tượng (OOP) trong Python. OOP là một mô hình lập trình (programming paradigm) mạnh mẽ, giúp chúng ta tổ chức code một cách logic, dễ quản lý và tái sử dụng. 

**Tại sao OOP quan trọng với Fintech?**

Trong lĩnh vực Fintech, chúng ta làm việc với các đối tượng phức tạp trong thế giới thực như:
* **Khách hàng (Customer)**
* **Tài khoản ngân hàng (Bank Account)**
* **Giao dịch (Transaction)**
* **Cổ phiếu (Stock):** Giấy tờ chứng nhận quyền sở hữu một phần của một công ty.
* **Trái phiếu (Bond):** Một loại chứng khoán nợ, về cơ bản là một khoản vay mà nhà đầu tư cho một tổ chức (chính phủ hoặc doanh nghiệp) vay.
* **Danh mục đầu tư (Portfolio):** Một tập hợp các khoản đầu tư khác nhau (cổ phiếu, trái phiếu, bất động sản...) do một cá nhân hoặc tổ chức nắm giữ.

OOP cho phép chúng ta "mô hình hóa" các đối tượng này thành các đối tượng trong code, với các thuộc tính (dữ liệu) và hành vi (phương thức) riêng. Điều này giúp code trở nên trực quan, dễ hiểu và dễ dàng mở rộng khi các sản phẩm tài chính ngày càng phức tạp.

## 1. Khái niệm cốt lõi: Lớp (Class) và Đối tượng (Object)

* **Lớp (Class):** Là một "bản thiết kế" hoặc "khuôn mẫu" để tạo ra các đối tượng. Nó định nghĩa các thuộc tính (attributes - đặc điểm) và phương thức (methods - hành vi) mà các đối tượng thuộc lớp đó sẽ có.
    * *Ví dụ:* Lớp `Nguoi` có thể định nghĩa các thuộc tính như `ten`, `tuoi` và phương thức `chao_hoi()`.
    * *Tưởng tượng:* Nếu `Class` là một khuôn làm bánh, thì `Object` chính là những chiếc bánh được tạo ra từ khuôn đó.

* **Đối tượng (Object):** Là một "thể hiện" (instance) cụ thể của một lớp. Mỗi đối tượng có dữ liệu (thuộc tính) riêng nhưng chia sẻ cùng cấu trúc và hành vi (phương thức) được định nghĩa bởi lớp.
    * *Ví dụ:* Dựa trên lớp `Nguoi`, ta có thể tạo ra đối tượng `sinh_vien_A` có `ten` là "An", `tuoi` là 20 và đối tượng `sinh_vien_B` có `ten` là "Bình", `tuoi` là 21.

### Ví dụ đầu tiên: Xây dựng lớp `BankAccount`

#### Bước 1: Thiết kế lớp `BankAccount`

Trước khi viết code, hãy cùng suy nghĩ về một tài khoản ngân hàng trong thực tế. Nó cần những thông tin gì và có thể làm được những gì?

**1. Các thông tin cần lưu trữ (Thuộc tính - Attributes):**
- `owner_name`: Tên của chủ tài khoản (kiểu chuỗi).
- `account_number`: Số tài khoản (kiểu chuỗi, vì có thể có số 0 ở đầu).
- `balance`: Số dư hiện tại (kiểu số).
- `currency`: Đơn vị tiền tệ, ví dụ: 'VND', 'USD' (kiểu chuỗi).

**2. Các hành động có thể thực hiện (Phương thức - Methods):**
- `deposit(amount)`: Gửi một khoản tiền `amount` vào tài khoản.
- `withdraw(amount)`: Rút một khoản tiền `amount` khỏi tài khoản.
- `check_balance()`: Kiểm tra (in ra) thông tin và số dư của tài khoản.

Bản thiết kế này sẽ là kim chỉ nam để chúng ta viết code cho lớp `BankAccount` ở bước tiếp theo.

```python
# Bước 2: Viết code dựa trên thiết kế

# Từ khóa `class` được dùng để bắt đầu định nghĩa một lớp mới.
class BankAccount:
    # --- HÀM KHỞI TẠO (CONSTRUCTOR) ---
    # `__init__` là một "phương thức đặc biệt" (special method) được tự động gọi mỗi khi một đối tượng mới được tạo ra từ lớp.
    # Nhiệm vụ của nó là để khởi tạo các giá trị ban đầu cho thuộc tính của đối tượng.
    #
    # `self` là một tham số bắt buộc trong mọi phương thức của một lớp.
    # Nó đại diện cho chính đối tượng (instance) đang được thao tác.
    # Tưởng tượng `self` như là "tôi" hoặc "của tôi". Ví dụ: `self.owner_name` có nghĩa là "tên chủ tài khoản của tôi".
    def __init__(self, owner_name, account_number, balance=0, currency='VND'):
        # --- THUỘC TÍNH (ATTRIBUTES) ---
        # Đây là các biến thuộc về một đối tượng, lưu trữ dữ liệu của riêng đối tượng đó.
        # Dòng `self.owner_name = owner_name` có nghĩa là: 
        # "Gán giá trị của tham số `owner_name` truyền vào cho thuộc tính `owner_name` CỦA TÔI (của self)".
        self.owner_name = owner_name
        self.account_number = account_number
        self.balance = balance
        self.currency = currency
        print(f"Tài khoản {self.account_number} của {self.owner_name} đã được tạo thành công!")

    # --- PHƯƠNG THỨC (METHODS) ---
    # Phương thức về cơ bản là một hàm thuộc về một lớp. Nó định nghĩa các hành vi, hành động mà đối tượng có thể thực hiện.
    # Phương thức luôn cần `self` làm tham số đầu tiên để có thể truy cập và thay đổi dữ liệu của chính đối tượng đó.
    def deposit(self, amount):
        """Đây là một Docstring - chuỗi tài liệu mô tả phương thức này làm gì."""
        if amount > 0:
            self.balance += amount # Truy cập và thay đổi thuộc tính `balance` của chính đối tượng
            print(f"Gửi thành công {amount} {self.currency}. Số dư mới: {self.balance} {self.currency}")
        else:
            print("Số tiền gửi vào phải lớn hơn 0.")

    def withdraw(self, amount):
        """Phương thức rút tiền từ tài khoản."""
        if amount <= 0:
            print("Số tiền rút ra phải lớn hơn 0.")
        elif self.balance >= amount:
            self.balance -= amount
            print(f"Rút thành công {amount} {self.currency}. Số dư mới: {self.balance} {self.currency}")
        else:
            print("Giao dịch thất bại. Số dư không đủ.")

    def check_balance(self):
        """Phương thức kiểm tra số dư."""
        print(f"Tài khoản: {self.account_number}\nChủ tài khoản: {self.owner_name}\nSố dư: {self.balance} {self.currency}")
```

#### Tạo và sử dụng các đối tượng `BankAccount`

```python
# Đây là quá trình "khởi tạo đối tượng" (instantiation) - tạo ra một thể hiện cụ thể từ lớp BankAccount.
# Khi dòng code này chạy, phương thức `__init__` của lớp BankAccount sẽ được tự động gọi.
account_A = BankAccount(owner_name="Nguyễn Văn A", account_number="0123456789", balance=10000000)

print("\n--- Thao tác trên tài khoản A ---")
# Sử dụng dấu chấm `.` để truy cập vào các thuộc tính hoặc gọi các phương thức của một đối tượng.
# `account_A.check_balance()` có nghĩa là: "Yêu cầu đối tượng `account_A` thực hiện hành động `check_balance` của nó".
account_A.check_balance()
account_A.deposit(5000000)
account_A.withdraw(2000000)


# Tạo một đối tượng khác. `account_B` và `account_A` độc lập với nhau, mỗi đối tượng có dữ liệu riêng.
account_B = BankAccount(owner_name="Trần Thị B", account_number="9876543210", currency="USD")

print("\n--- Thao tác trên tài khoản B ---")
account_B.deposit(500)
account_B.withdraw(1000)
account_B.check_balance()
```

Mở rộng lớp `BankAccount` đã tạo:
1.  Trong phương thức `__init__`, hãy thêm một thuộc tính mới là `transaction_history`, khởi tạo là một danh sách rỗng `[]`.
2.  Sửa đổi phương thức `deposit` và `withdraw`: Mỗi khi một giao dịch thành công, hãy thêm một chuỗi mô tả giao dịch vào `transaction_history`. 
    * Ví dụ: `f"Gửi tiền: +{amount}"` hoặc `f"Rút tiền: -{amount}"`.
3.  Tạo một phương thức mới tên là `print_history()` để in ra tất cả các giao dịch trong lịch sử.

## 2. Bốn Trụ Cột của OOP
OOP được xây dựng dựa trên 4 nguyên tắc chính. Hiểu rõ chúng sẽ giúp bạn viết code hiệu quả hơn.

### 2.1. Tính đóng gói (Encapsulation)
**Ý tưởng:** Gói gọn dữ liệu (thuộc tính) và các phương thức xử lý dữ liệu đó vào trong cùng một đối tượng. Mục đích là để **che giấu thông tin** (information hiding), hạn chế việc truy cập và thay đổi dữ liệu một cách tùy tiện từ bên ngoài, đảm bảo dữ liệu được xử lý đúng cách thông qua các phương thức đã định nghĩa.

**Trong Python:** Chúng ta thường quy ước các thuộc tính "private" (riêng tư) bằng cách thêm hai dấu gạch dưới `__` ở đầu tên. Điều này báo hiệu rằng thuộc tính này chỉ nên được sử dụng bên trong lớp.

**Ví dụ:** Chúng ta không muốn ai đó có thể thay đổi số dư một cách trực tiếp như `account_A.balance = 999999999`. Thay vào đó, họ phải dùng phương thức `deposit` hoặc `withdraw`.

#### Thiết kế lớp `BankAccountSecure`

**Vấn đề:** Ở lớp `BankAccount` trước, bất kỳ ai cũng có thể thay đổi số dư trực tiếp: `account_A.balance = -1000000`. Điều này rất nguy hiểm và làm mất tính toàn vẹn của dữ liệu.

**Mục tiêu thiết kế:** Chúng ta cần một phiên bản an toàn hơn của tài khoản ngân hàng, trong đó thuộc tính `balance` được bảo vệ.

**Giải pháp (Áp dụng Tính đóng gói):**
1.  **"Private" hóa thuộc tính `balance`:** Đổi tên `balance` thành `__balance`. Dấu `__` báo hiệu rằng nó là thuộc tính nội bộ, không nên bị truy cập từ bên ngoài.
2.  **Cung cấp phương thức truy cập an toàn:** Vì `__balance` đã bị ẩn, chúng ta cần tạo một phương thức công khai (public) để bên ngoài có thể *đọc* được giá trị của nó một cách có kiểm soát. Ta sẽ tạo phương thức `get_balance()`.
3.  **Giữ nguyên các phương thức public:** Các phương thức `deposit` và `withdraw` vẫn là public vì đó là cách hợp lệ duy nhất để thay đổi số dư.

Với thiết kế này, chúng ta kiểm soát hoàn toàn cách dữ liệu được truy cập và thay đổi.

```python
class BankAccountSecure:
    def __init__(self, owner_name, account_number, balance=0):
        self.owner_name = owner_name
        self.account_number = account_number
        # __balance là một thuộc tính được coi là "private".
        # Bằng cách thêm `__`, Python sẽ tự động đổi tên nó (name mangling) để tránh bị truy cập vô tình từ bên ngoài.
        self.__balance = balance 

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Gửi thành công {amount}. Số dư mới đã được cập nhật.")
        else:
            print("Số tiền gửi vào phải lớn hơn 0.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Rút thành công {amount}. Số dư mới đã được cập nhật.")
        else:
            print("Số tiền không hợp lệ hoặc số dư không đủ.")

    # Cung cấp một phương thức "public" (công khai) để đọc dữ liệu một cách an toàn.
    # Đây được gọi là một "getter method".
    def get_balance(self):
        return self.__balance
    
    def display_info(self):
        print(f"Tài khoản {self.account_number} của {self.owner_name}")

# Tạo đối tượng mới
secure_acc = BankAccountSecure("Lê An Toàn", "1122334455", 5000)

# Thử truy cập trực tiếp vào thuộc tính private -> sẽ gây lỗi!
try:
    print(secure_acc.__balance)
except AttributeError as e:
    print(f"Lỗi: {e}")
    print("==> Không thể truy cập `__balance` trực tiếp từ bên ngoài.")

# Phải truy cập thông qua phương thức public đã được cung cấp
print(f"Số dư hiện tại là: {secure_acc.get_balance()}")

# Mọi thay đổi về số dư đều phải thông qua các phương thức public, đảm bảo tính toàn vẹn dữ liệu
secure_acc.deposit(1000)
print(f"Số dư sau khi gửi tiền: {secure_acc.get_balance()}")
```

#### ⭐ Bài tập thực hành 2: Kiểm soát dữ liệu

Nâng cấp lớp `BankAccountSecure`:
1.  Thêm một phương thức `change_owner(new_name)`.
2.  Bên trong phương thức này, hãy kiểm tra xem `new_name` có phải là một chuỗi (`str`) và không rỗng (ví dụ: `len(new_name) > 0`) hay không. 
3.  Nếu hợp lệ, hãy cập nhật `self.owner_name`. Nếu không, hãy in ra một thông báo lỗi. 
Điều này thể hiện việc kiểm soát cách dữ liệu được thay đổi, một phần quan trọng của tính đóng gói.

### 2.2. Tính kế thừa (Inheritance)

**Ý tưởng:** Cho phép một lớp mới (gọi là **lớp con** - `child class`) thừa hưởng lại các thuộc tính và phương thức từ một lớp đã có (gọi là **lớp cha** - `parent class`). Lớp con có thể sử dụng lại code của lớp cha, mở rộng (thêm thuộc tính/phương thức mới) hoặc ghi đè (thay đổi hành vi của phương thức cũ).

**Lợi ích:** Tái sử dụng code, tránh lặp lại code và tạo ra một hệ thống phân cấp logic (mối quan hệ "là một loại của" - "is-a").

**Ví dụ trong Fintech:** Một `SavingAccount` (Tài khoản tiết kiệm) *là một loại của* `Account`. Nó có tất cả đặc điểm của một tài khoản thông thường, cộng thêm `lãi suất`.

#### Thiết kế các lớp `Account`, `SavingAccount` và `CheckingAccount`

**Vấn đề:** Trong thực tế, có nhiều loại tài khoản ngân hàng: tài khoản tiết kiệm, tài khoản thanh toán... Mỗi loại có đặc điểm chung nhưng cũng có những điểm riêng. Viết một lớp riêng cho mỗi loại từ đầu sẽ rất lặp code.

**Mục tiêu thiết kế:** Xây dựng một hệ thống các lớp tài khoản có thể tái sử dụng code và dễ dàng mở rộng.

**Giải pháp (Áp dụng Tính kế thừa):**
1.  **Lớp cha `Account`:** Tạo một lớp cha chung, kế thừa từ `BankAccountSecure` để có sẵn các tính năng cơ bản và an toàn.
2.  **Lớp con `SavingAccount`:** 
    - **Quan hệ:** Một `SavingAccount` *là một loại* `Account`.
    - **Kế thừa:** Thừa hưởng tất cả thuộc tính và phương thức từ `Account`.
    - **Mở rộng:** Thêm thuộc tính mới là `interest_rate` (lãi suất) và phương thức mới là `calculate_interest()`.
3.  **Lớp con `CheckingAccount`:** 
    - **Quan hệ:** Một `CheckingAccount` *là một loại* `Account`.
    - **Kế thừa:** Thừa hưởng tất cả từ `Account`.
    - **Mở rộng:** Thêm thuộc tính mới là `overdraft_limit` (hạn mức thấu chi). *Giải thích: Thấu chi là việc chi tiêu vượt quá số tiền có trong tài khoản, trong một giới hạn được ngân hàng cho phép.*
    - **Ghi đè (Override):** Thay đổi hành vi của phương thức `withdraw` để cho phép rút quá số dư (thấu chi) trong giới hạn cho phép. 

```python
# Để đơn giản, ta định nghĩa lại lớp cha ở đây
class Account(BankAccountSecure): # Cú pháp `(BankAccountSecure)` có nghĩa là lớp Account kế thừa từ BankAccountSecure.
    def __init__(self, owner_name, account_number, balance=0):
        # `super()` là một hàm đặc biệt dùng để gọi phương thức từ lớp cha.
        # Ở đây, ta gọi `__init__` của lớp cha (BankAccountSecure) để nó thực hiện việc khởi tạo các thuộc tính cơ bản.
        super().__init__(owner_name, account_number, balance)
        print("Lớp Account cơ bản được khởi tạo.")

# Lớp con (Child Class)
class SavingAccount(Account):
    def __init__(self, owner_name, account_number, balance=0, interest_rate=0.05):
        super().__init__(owner_name, account_number, balance)
        self.interest_rate = interest_rate # Thêm một thuộc tính mới chỉ có ở SavingAccount
        print("Lớp SavingAccount được khởi tạo.")
        
    # Thêm một phương thức mới
    def calculate_interest(self):
        interest = self.get_balance() * self.interest_rate
        print(f"Tiền lãi dự kiến: {interest}")
        return interest
        
# Lớp con khác
class CheckingAccount(Account):
    def __init__(self, owner_name, account_number, balance=0, overdraft_limit=5000000):
        super().__init__(owner_name, account_number, balance)
        self.overdraft_limit = overdraft_limit
        print("Lớp CheckingAccount được khởi tạo.")
        
    # --- GHI ĐÈ PHƯƠNG THỨC (METHOD OVERRIDING) ---
    # Lớp con định nghĩa lại một phương thức đã có ở lớp cha để cung cấp một hành vi khác, chuyên biệt hơn.
    # Ở đây, `CheckingAccount` có cách rút tiền khác với `Account` thông thường.
    def withdraw(self, amount):
        """Cho phép rút tiền ngay cả khi số dư không đủ, trong giới hạn thấu chi."""
        if amount > 0 and (self.get_balance() + self.overdraft_limit >= amount):
            # Logic rút tiền mới, phức tạp hơn
            print("Đang xử lý rút tiền cho tài khoản thanh toán...")
            # Gọi phương thức withdraw của lớp cha để thực hiện logic trừ tiền cơ bản nếu có thể
            super().withdraw(amount)
        else:
            print("Số tiền không hợp lệ hoặc vượt quá hạn mức thấu chi.")
```

```python
print("--- Tạo tài khoản tiết kiệm ---")
my_saving = SavingAccount("Phạm Thịnh Vượng", "88889999", 200000000, 0.06)
# `my_saving` có thể gọi `display_info()` dù phương thức này không được định nghĩa trong lớp `SavingAccount`.
# Đó là vì nó đã được kế thừa từ lớp cha `Account`.
my_saving.display_info()
print(f"Số dư: {my_saving.get_balance()}")
my_saving.calculate_interest() # Đây là phương thức riêng của SavingAccount

print("\n--- Tạo tài khoản thanh toán ---")
my_checking = CheckingAccount("Nguyễn Năng Động", "77778888", 1000000)
my_checking.display_info()
print(f"Số dư: {my_checking.get_balance()}")
# Khi gọi `withdraw`, phiên bản của `CheckingAccount` sẽ được thực thi, chứ không phải của lớp cha.
my_checking.withdraw(3000000)
```

### 2.3. Tính đa hình (Polymorphism)

**Ý tưởng:** "Poly" có nghĩa là "nhiều", "morph" có nghĩa là "hình dạng". Đa hình có nghĩa là "nhiều hình dạng". Trong OOP, nó thể hiện khả năng các đối tượng thuộc các lớp khác nhau có thể phản hồi cùng một lời gọi phương thức, nhưng theo cách riêng của chúng. 

**Lợi ích:** Giúp viết code tổng quát hơn. Một đoạn code có thể làm việc với nhiều loại đối tượng khác nhau mà không cần biết chi tiết về lớp của chúng, miễn là chúng có chung một giao diện (interface) - ví dụ như cùng có một phương thức tên là `calculate_value`.

**Ví dụ trong Fintech:** Giả sử chúng ta có một danh mục đầu tư bao gồm nhiều loại tài sản khác nhau. Ta có thể duyệt qua tất cả tài sản và chỉ cần gọi `asset.calculate_value()` mà không cần dùng `if...else` để kiểm tra `asset` là Cổ phiếu, Trái phiếu hay Bất động sản.

#### Thiết kế các lớp tài sản cho danh mục đầu tư

**Vấn đề:** Một danh mục đầu tư có thể chứa nhiều loại tài sản: `Stock`, `Bond`, `RealEstate`... Mỗi loại có cách tính giá trị khác nhau. Làm thế nào để tính tổng giá trị danh mục một cách linh hoạt, dễ dàng thêm loại tài sản mới trong tương lai?

**Mục tiêu thiết kế:** Viết một đoạn code có thể tính tổng giá trị của một danh sách chứa nhiều loại tài sản khác nhau mà không cần câu lệnh `if-else` để kiểm tra loại tài sản.

**Giải pháp (Áp dụng Tính đa hình):**
1.  **Thiết lập một "Giao diện chung" (Common Interface):** Quy ước rằng tất cả các lớp tài sản, dù là loại gì, đều phải có một phương thức với tên gọi giống hệt nhau để tính giá trị. Chúng ta sẽ chọn tên là `calculate_value()`.
2.  **Triển khai riêng cho từng lớp:**
    - Lớp `Stock`: triển khai `calculate_value()` bằng cách lấy `số lượng * giá thị trường`.
    - Lớp `Bond`: triển khai `calculate_value()` bằng cách lấy `mệnh giá`. *Giải thích: Mệnh giá (face value) là giá trị ban đầu của trái phiếu, số tiền sẽ được trả lại cho nhà đầu tư khi đáo hạn.*
    - Lớp `RealEstate`: triển khai `calculate_value()` bằng cách lấy `giá trị thẩm định`. *Giải thích: Giá trị thẩm định (appraised value) là giá trị ước tính của một tài sản bởi một chuyên gia.*
3.  **Sử dụng tính đa hình:** Khi duyệt qua danh mục, chúng ta chỉ cần gọi `asset.calculate_value()` trên mỗi đối tượng. Python sẽ tự động biết phải chạy phiên bản `calculate_value()` nào (của Stock, Bond hay RealEstate) dựa trên loại đối tượng thực tế của `asset` tại thời điểm đó.

```python
class Stock:
    def __init__(self, ticker, quantity, market_price):
        self.ticker = ticker
        self.quantity = quantity
        self.market_price = market_price
    
    def calculate_value(self):
        print(f"(Tính giá trị Cổ phiếu: số lượng * giá thị trường)")
        return self.quantity * self.market_price
    
    # `__str__` là một phương thức đặc biệt khác. Nó định nghĩa chuỗi ký tự sẽ được hiển thị khi ta print() một đối tượng.
    def __str__(self):
        return f"Cổ phiếu {self.ticker}"

class Bond:
    def __init__(self, name, face_value):
        self.name = name
        self.face_value = face_value
    
    def calculate_value(self):
        print(f"(Tính giá trị Trái phiếu: dựa trên mệnh giá)")
        return self.face_value

    def __str__(self):
        return f"Trái phiếu {self.name}"

class RealEstate:
    def __init__(self, address, appraised_value):
        self.address = address
        self.appraised_value = appraised_value
    
    def calculate_value(self):
        print(f"(Tính giá trị BĐS: dựa trên giá trị thẩm định)")
        return self.appraised_value

    def __str__(self):
        return f"Bất động sản tại {self.address}"

portfolio = [
    Stock("VNM", 1000, 75000),
    Bond("Chính phủ 10 năm", 100000000),
    RealEstate("Quận 1, TPHCM", 20000000000)
]

total_portfolio_value = 0
for asset in portfolio:
    # Đây chính là tính đa hình!
    # Cùng một lời gọi `asset.calculate_value()` nhưng Python sẽ tự động chọn đúng phương thức để chạy
    # tùy thuộc vào `asset` lúc đó là Stock, Bond hay RealEstate.
    asset_value = asset.calculate_value()
    print(f"Giá trị của {asset}: {asset_value:,.0f} VND\n")
    total_portfolio_value += asset_value

print(f"==> Tổng giá trị danh mục đầu tư: {total_portfolio_value:,.0f} VND")
```

### 2.4. Tính trừu tượng (Abstraction)

**Ý tưởng:** Che giấu sự phức tạp của việc triển khai bên trong và chỉ hiển thị các chức năng cần thiết ra bên ngoài. Nó giúp người dùng tập trung vào việc *sử dụng* đối tượng thay vì phải quan tâm nó được *xây dựng* như thế nào.

**Lợi ích:** Giảm độ phức tạp và tăng tính linh hoạt. Chúng ta có thể thay đổi cách triển khai bên trong một lớp mà không làm ảnh hưởng đến các đoạn code khác đang sử dụng lớp đó.

Trong lập trình, chúng ta thường sử dụng **Lớp trừu tượng (Abstract Class)**. Đây là một lớp "chưa hoàn chỉnh", không thể tạo đối tượng trực tiếp từ nó. Nó định nghĩa ra một "bản hợp đồng" hay một bộ khung các phương thức mà các lớp con kế thừa từ nó *bắt buộc* phải triển khai.

#### Thiết kế Lớp trừu tượng `FinancialInstrument`

**Vấn đề:** Ở ví dụ về tính đa hình, việc tất cả các lớp tài sản đều có phương thức `calculate_value` chỉ là một *quy ước ngầm*. Nếu một lập trình viên khác tạo ra một lớp `CryptoCurrency` mới nhưng quên không định nghĩa phương thức này, chương trình sẽ bị lỗi khi chạy vòng lặp tính tổng giá trị. 

**Mục tiêu thiết kế:** Cần một cách để *bắt buộc* mọi lớp công cụ tài chính trong hệ thống của chúng ta phải tuân theo một cấu trúc chung. Cụ thể, bất kỳ công cụ tài chính nào cũng phải có một phương thức để lấy giá trị hiện tại.

**Giải pháp (Áp dụng Tính trừu tượng):**
1.  **Tạo Lớp trừu tượng `FinancialInstrument`:** Đây sẽ là "bản hợp đồng" chung. Nó không phải là một công cụ tài chính cụ thể, mà là một khái niệm chung.
2.  **Định nghĩa một Phương thức trừu tượng:** Bên trong `FinancialInstrument`, ta định nghĩa một phương thức là `get_current_value()` và đánh dấu nó là `@abstractmethod`. Phương thức này không có code, nó chỉ tồn tại để ra lệnh: "Ai kế thừa từ tôi thì bắt buộc phải tự viết code cho phương thức này".
3.  **Các lớp con phải tuân thủ:** Khi các lớp cụ thể như `Stock` hay `MutualFund` (*Quỹ tương hỗ - một loại hình đầu tư tập thể*) kế thừa từ `FinancialInstrument`, Python sẽ báo lỗi nếu chúng không định nghĩa (implement) phương thức `get_current_value()`. Điều này đảm bảo tính nhất quán và đúng đắn cho toàn bộ hệ thống.

```python
# `abc` là một module có sẵn của Python, viết tắt của Abstract Base Classes.
# Ta import `ABC` để tạo lớp trừ tượng và `abstractmethod` để đánh dấu phương thức trừu tượng.
from abc import ABC, abstractmethod

# Một lớp trừu tượng thường kế thừa từ ABC.
class FinancialInstrument(ABC):
    def __init__(self, name):
        self.name = name

    # `@abstractmethod` là một "decorator". Nó giống như một nhãn dán, đánh dấu phương thức ngay bên dưới nó là trừu tượng.
    # Phương thức trừu tượng không có code triển khai (chỉ có `pass`).
    @abstractmethod
    def get_current_value(self):
        pass

    def display_info(self):
        print(f"Đây là một công cụ tài chính tên: {self.name}")

# Lớp con triển khai lớp trừu tượng
class Stock(FinancialInstrument):
    def __init__(self, name, quantity, price):
        super().__init__(name)
        self.quantity = quantity
        self.price = price

    # Lớp Stock BẮT BUỘC phải định nghĩa phương thức `get_current_value`,
    # vì nó đã kế thừa từ lớp trừu tượng `FinancialInstrument`.
    def get_current_value(self):
        return self.quantity * self.price

class MutualFund(FinancialInstrument):
    def __init__(self, name, shares, nav):
        super().__init__(name)
        self.shares = shares
        self.nav = nav # Net Asset Value (Giá trị tài sản ròng trên một chứng chỉ quỹ)
    
    def get_current_value(self):
        return self.shares * self.nav

# Thử tạo đối tượng từ lớp trừu tượng -> sẽ gây lỗi!
# Vì nó là bản thiết kế chưa hoàn chỉnh.
try:
    instrument = FinancialInstrument("Chung chung")
except TypeError as e:
    print(f"Lỗi: {e}")

# Tạo đối tượng từ các lớp con đã triển khai đầy đủ thì hoàn toàn bình thường.
stock_fpt = Stock("FPT", 500, 120000)
fund_vcb = MutualFund("Quỹ VCBF", 10000, 25000)

assets = [stock_fpt, fund_vcb]
for asset in assets:
    asset.display_info()
    print(f"Giá trị hiện tại: {asset.get_current_value():,.0f} VND\n")
```
